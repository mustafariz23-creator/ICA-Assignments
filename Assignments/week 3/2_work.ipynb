```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Administrator\Downloads\teaching_prices_messy.csv")
df['price'] = pd.to_numeric(df['price'].astype(str).str.replace(',', ''), errors='coerce')
df = df.dropna(subset=['price']).copy()

for item_name, group in df.groupby('item'):
    group = group.sort_values('date')
    mean_val = group['price'].mean()
    median_val = group['price'].median()
    first_val = group['price'].iloc[0]
    last_val = group['price'].iloc[-1]
    pct_change = ((last_val - first_val) / first_val) * 100

    print(f"Item: {item_name}")
    print(f"  Mean: {mean_val:.2f} | Median: {median_val:.2f}")
    print(f"  First-to-last % change: {pct_change:.2f}% ({first_val} -> {last_val})\n")
```

```python
sample_item = df['item'].iloc[0]
sub = df[df['item'] == sample_item].copy()

print("With all values:")
print("  Mean:", round(sub['price'].mean(), 2), "Median:", round(sub['price'].median(), 2))
sub_no_max = sub[sub['price'] < sub['price'].max()]
print("Without highest value:")
print("  Mean:", round(sub_no_max['price'].mean(), 2), "Median:", round(sub_no_max['price'].median(), 2))
```
