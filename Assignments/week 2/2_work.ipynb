applying Python code to fix missing values, correct item casing, drop duplicate entries, and clean up the text formatting in the price column.

```python
import pandas as pd

df_raw = pd.read_csv('teaching_prices_messy.csv')
df_clean = df_raw.copy()

print("Original shape:", df_raw.shape)
df_clean.head()
```

```python

df_clean = df_raw.copy()
df_clean = df_clean.drop_duplicates()
df_clean['price'] = df_clean['price'].astype(str).str.replace(',', '')
df_clean['price'] = pd.to_numeric(df_clean['price'])

print(df_clean.isnull().sum())
```
